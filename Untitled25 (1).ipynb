```python
# Load the passport photo

img = cv2.imread("litya .jpeg")

# Check if the image was loaded correctly
if img is None:
    print("Image not found. Check the file name and path.")
else:
    print("Passport photo loaded successfully!")

# Convert BGR to RGB for displaying with Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the original image
plt.figure(figsize=(5, 5))
plt.imshow(img_rgb)
plt.axis("off")
plt.title("Original Passport Photo")
plt.show()
```

```python
# Load the sunglasses image
# IMREAD_UNCHANGED keeps the transparent (alpha) background

sunglasses = cv2.imread("sunglass.jpg", cv2.IMREAD_UNCHANGED)

# Check if the sunglasses image was loaded
if sunglasses is None:
    print("Sunglasses image not found. Check the file name and path.")
else:
    print("Sunglasses image loaded successfully!")

# Display the sunglasses image
plt.figure(figsize=(6, 3))

if sunglasses.shape[2] == 4:
    # Convert BGRA to RGBA for Matplotlib
    sunglasses_rgb = cv2.cvtColor(sunglasses, cv2.COLOR_BGRA2RGBA)
    plt.imshow(sunglasses_rgb)
else:
    # Convert BGR to RGB
    sunglasses_rgb = cv2.cvtColor(sunglasses, cv2.COLOR_BGR2RGB)
    plt.imshow(sunglasses_rgb)

plt.axis("off")
plt.title("Sunglasses")
plt.show()
```

```python
# Check the size of the passport photo

height, width = img.shape[:2]

print("Image width :", width)
print("Image height:", height)

# Display the image with a coordinate grid
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)

plt.xticks(range(0, width, max(1, width // 10)))
plt.yticks(range(0, height, max(1, height // 10)))
plt.grid()

plt.title("Find the Eye Region Coordinates")
plt.show()
```

```python
# Eye region coordinates

x = 50
y = 480
w = 350
h = 130

# Crop the eye region
eye_region = img[y:y+h, x:x+w]

# Display the selected eye region
eye_region_rgb = cv2.cvtColor(eye_region, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 3))
plt.imshow(eye_region_rgb)
plt.axis("off")
plt.title("Selected Eye Region")
plt.show()
```

```python
sunglasses_resized = cv2.resize(sunglasses, (w, h))

gray_sunglasses = cv2.cvtColor(
    sunglasses_resized,
    cv2.COLOR_BGR2GRAY
)

_, mask = cv2.threshold(
    gray_sunglasses,
    300,
    255,
    cv2.THRESH_BINARY_INV
)
```

```python
# Get eye region
eye_region = img[y:y+h, x:x+w]

# Convert mask to alpha
alpha = mask.astype(float) / 255.0

# Make 3-channel alpha
alpha = cv2.merge([alpha, alpha, alpha])

# Convert to float
eye_float = eye_region.astype(float)
sunglasses_float = sunglasses_resized.astype(float)

# Blend
blended = (
    sunglasses_float * alpha +
    eye_float * (1 - alpha)
)

# Convert back to uint8
blended = np.clip(blended, 0, 255).astype(np.uint8)

# Put back into image
result = img.copy()
result[y:y+h, x:x+w] = blended

# Display
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(7, 9))
plt.imshow(result_rgb)
plt.axis("off")
plt.title("Final Image with Sunglasses")
plt.show()
```
